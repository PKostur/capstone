```python
from datasets import load_dataset, get_dataset_config_names
import pandas as pd
import spacy
from gensim.utils import simple_preprocess
import json
from collections import Counter
```

```python
dataset = load_dataset("cjvt/ParlaMint3", 'gb', trust_remote_code=True, verification_mode="no_checks")
dataset_df = pd.DataFrame(dataset['train'])
dataset_df = dataset_df[['Speaker_party_name', 'text']]
```

```python
!python -m spacy download en_core_web_sm
```

```python
nlp = spacy.load("en_core_web_sm", disable=["parser", "ner"])
```

```python
def preprocess_text(text):
    tokens = simple_preprocess(text, deacc=True)  
    doc = nlp(" ".join(tokens))  
    lemmas = [token.lemma_ for token in doc if token.is_alpha and not token.is_stop] 

    return lemmas
```

```python
chunk_size = 10000
processed_chunks = []

for i in range(0, len(dataset_df), chunk_size):
    print(f"Processing chunk {i} to {i + chunk_size}...")
    chunk = dataset_df.iloc[i:i + chunk_size].copy()
    chunk['tokens'] = chunk['text'].apply(preprocess_text)  
    processed_chunks.append(chunk)

df = pd.concat(processed_chunks, ignore_index=True)

df.to_json('C:/Users/007pe/Downloads/dataset_preprocessed.json', orient='records', lines=True)
```

Here is where I do some analysis of the text data, such as finding the most frequent terms or the most frequent bi-grams(two-word combinations)

```python
df = pd.read_json('C:/Users/007pe/Downloads/dataset_preprocessed.json', lines=True)
```

```python
all_words = [word for tokens in df['tokens'] for word in tokens]
word_freq = Counter(all_words)

top_words = word_freq.most_common(20)

word_freq_df = pd.DataFrame(top_words, columns=['Word', 'Frequency'])
print("Top 20 Most Frequent Words:")
print(word_freq_df)
```

```python
party_tokens = df.groupby('Speaker_party_name')['tokens'].agg(sum)  

party_word_freq = {party: Counter(tokens) for party, tokens in party_tokens.items()}

top_n = 20
for party in party_tokens.index:
    print(f"\nTop {top_n} words for {party}:")
    top_words = party_word_freq[party].most_common(top_n)
    print(pd.DataFrame(top_words, columns=['Word', 'Frequency']))
```

```python
df['word_count'] = df['tokens'].apply(len)

word_count_stats = df['word_count'].describe()
print("Text Length Distribution (Word Count per Speech):")
print(word_count_stats)
```

```python
unique_words = len(set(all_words))
print(f"Total Unique Words: {unique_words}")
```

```python
!pip install nltk
import nltk
from nltk import bigrams

# Generate bigrams for each speech
df['bigrams'] = df['tokens'].apply(lambda tokens: list(bigrams(tokens)))

# Flatten and count bigrams
all_bigrams = [bigram for bigrams_list in df['bigrams'] for bigram in bigrams_list]
bigram_freq = Counter(all_bigrams)

# Get the top 10 bigrams
top_bigrams = bigram_freq.most_common(10)

# Create a table
bigram_freq_df = pd.DataFrame(top_bigrams, columns=['Bigram', 'Frequency'])
bigram_freq_df['Bigram'] = bigram_freq_df['Bigram'].apply(lambda x: ' '.join(x))
print("Top 10 Most Frequent Bigrams:")
print(bigram_freq_df)
```
