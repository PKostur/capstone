```python
import pandas as pd
import numpy as np

from scipy.sparse import vstack
from sklearn.svm import SVC

from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_json('C:/Users/007pe/Downloads/tdidf_embeddings.json', lines=True)
```

```python
df = df.dropna(subset=['Speaker_party_name'])
print("DataFrame after dropping NaN in 'Speaker_party_name':")
print(df)
```

```python
print(df)
df['Speaker_party_name'] = df['Speaker_party_name'].dropna()
print(df)
# print(df['Speaker_party_name'])
# df2 = df.copy(deep=True)
```

```python
X = np.vstack(df['embedding'].values)
y = df['Speaker_party_name']
```

```python
print(y.value_counts())
y = y.dropna()
print(y.value_counts())
```

```python
encoder = LabelEncoder()

y = encoder.fit_transform(y)  # Convert labels to integers
num_classes = len(encoder.classes_) 
print(encoder.classes_)
print(num_classes)
```

```python
print(len(y))
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0, stratify=y)
```

```python
svm_classifier = SVC()
svm_classifier.fit(X_train, y_train)
```

```python
import joblib

svm_classifier_path = 'C:/Users/007pe/Downloads/svm_classifier_model.pkl'
# joblib.dump(svm_classifier, svm_classifier_path)
svm_classifier = joblib.load(svm_classifier_path)
```

```python
%%time
print("Making predictions...")
predictions = svm_classifier.predict(X_test)

# Evaluate
print("\nClassification Report:")
print(classification_report(y_test, predictions, zero_division=0))
print(f"Accuracy: {accuracy_score(y_test, predictions):.4f}")
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, predictions))
```
